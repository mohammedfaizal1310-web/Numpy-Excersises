```python
##How to find common values between two arrays
import numpy as np
arr1=[1,2,3,4,5]
arr2=[4,5,6,7,8]
common_values=np.intersect1d(arr1,arr2)
print(common_values)
```

```python
###How to get the dates of yesterday, today and tomorrow
import numpy as np
today=np.datetime64('today','D')
yesterday=today-np.timedelta64('1','D')
tommorrow=today+np.timedelta64('1','D')
print(yesterday,today,tommorrow)
```

```python
##How to get all the dates corresponding to the month of July 2016
import numpy as np
july_2016=np.arange('2016-07-01', '2016-08-01', dtype='datetime64[D]')
print(july_2016)
```

```python
##How to compute ((A+B)*(-A/2)) in place (without copy)
import numpy as np
A=np.ones(4)*3
B=np.ones(4)*2
np.add(A,B,out=B)
np.divide(A,-2,out=A)
np.multiply(B,A,out=A)
print(A)
```

```python
## Extract the integer part of a random array of positive numbers using 4 different methods
import numpy as np
z=np.random.uniform(0,10,10)
print(z.astype(int))
print(np.trunc(z))
print(z//1)
print(np.floor(z))
```

```python
## Create a 5x5 matrix with row values ranging from 0 to 4
import numpy as np
mat=np.array([0,1,2,3,4] , dtype=int)
mat=np.tile(mat,(5,1))
print(mat)
```

```python
##Consider a generator function that generates 10 integers and use it to build an array
import numpy as np
def genrate():
  for i in range(10):
    yield i
arr=np.fromiter(genrate(),dtype=int)
print(arr)
```
