```python
##Create a custom dtype that describes a color as four unsigned bytes (RGBA)
import numpy as np
rgba_dtype=np.dtype([("red",np.uint8),("green",np.uint8),("blue",np.uint8),("alpha",np.uint8)])
color=np.array([(255,0,0,255), (0,255,0,255),(0,0,255,255),(255,255,0,128)],dtype=rgba_dtype)
print(color)
```

```python
##Multiply a 5x3 matrix by a 3x2 matrix (real matrix product)
import numpy as np
mat1=np.ones((5,3),dtype=np.int32)
mat2=np.zeros((3,2),dtype=np.int32)
print(np.dot(mat1,mat2))
```

```python
## Given a 1D array, negate all elements which are between 3 and 8, in place
import numpy as np
arr=np.arange(11)
print(arr)
arr[(arr>3) & (arr<8)]*=-1
print(arr)
```

```python
##How to round away from zero a float array
import numpy as np
z=np.random.rand(10)
result=np.ceil(np.abs(z)*np.sign(z))
print(result)
```
